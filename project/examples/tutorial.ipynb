## 📚 Section 2: Understanding Key Concepts

### What is Telemetry?

**Telemetry** is the automatic measurement and transmission of data from remote sources. In the context of LLMs:
- It tracks every API call you make
- Records performance metrics (latency, tokens, costs)
- Enables data-driven optimization

### Why Do We Need Model Monitoring?

LLM applications face unique challenges:
1. **Variable Performance**: Response quality can vary
2. **Unpredictable Costs**: Token usage affects pricing
3. **Latency Issues**: Response time impacts user experience
4. **Error Rates**: APIs can fail or return poor results

### What is Prompt Optimization?

The same question can be asked in different ways:
- "Explain AI" → Generic
- "Explain AI in simple terms for a beginner" → Better
- "You are an expert teacher. Explain AI concepts to a complete beginner using analogies and simple language." → Best

**Prompt optimization** finds the best way to phrase requests for:
- ✅ Better quality responses
- ✅ Lower costs (fewer tokens)
- ✅ Faster responses
- ✅ Higher success rates

## 📊 Section 4: Building a Telemetry Logger

The **Telemetry Logger** is the heart of our system. It wraps around LLM API calls and automatically collects metrics.

### Key Metrics We Track:

| Metric | Description | Why It Matters |
|--------|-------------|----------------|
| **Latency** | Time from request to response | User experience |
| **Input Tokens** | Tokens in the prompt | Cost calculation |
| **Output Tokens** | Tokens in response | Cost calculation |
| **Total Cost** | API cost in USD | Budget management |
| **Error Rate** | % of failed requests | Reliability |
| **Quality Score** | Response quality (0-1) | Effectiveness |

### How It Works:

```python
with tracker.track_request(...) as ctx:
    # Your API call here
    response = call_llm()
    
    # Set metrics
    ctx.set_tokens(input, output)
    ctx.set_response(response)
```

Let's see it in action!

````python
def simulate_llm_call(model: str, prompt: str, simulate_response: str):
    """
    Simulates an LLM API call with telemetry tracking.
    In production, replace this with actual OpenAI/Anthropic calls.
    """
    with tracker.track_request(
        model_name=model,
        provider="openai",
        prompt_text=prompt,
        tags=["tutorial", "demo"]
    ) as ctx:
        # Simulate network latency
        time.sleep(random.uniform(0.1, 0.3))
        
        # Calculate token usage (rough estimate)
        input_tokens = int(len(prompt.split()) * 1.3)
        output_tokens = int(len(simulate_response.split()) * 1.3)
        
        # Set metrics
        ctx.set_tokens(input_tokens, output_tokens)
        ctx.set_response(simulate_response)
        ctx.set_quality_score(random.uniform(0.75, 0.95))
        ctx.set_user_feedback("positive" if random.random() < 0.8 else "negative")
        
        return simulate_response

# Example usage
prompt = "Explain what machine learning is in one sentence."
response = simulate_llm_call(
    model="gpt-3.5-turbo",
    prompt=prompt,
    simulate_response="Machine learning is a branch of AI that enables computers to learn from data and improve their performance without being explicitly programmed."
)

print("📝 Prompt:", prompt)
print("🤖 Response:", response)
print("✅ Telemetry data automatically captured and stored!")
````

## 🧪 Section 8: A/B Testing Framework

**A/B Testing** is crucial for data-driven prompt optimization. Instead of guessing which prompt is better, we:

1. Run multiple versions simultaneously
2. Split traffic between them (e.g., 33% each)
3. Collect performance data
4. Statistically determine the winner

### Traffic Splitting:

```
Control (33%) ────────> Version 1
                         
Variant A (33%) ───────> Version 2
                         
Variant B (34%) ───────> Version 3
```

Let's run an A/B test!

````python
print("🧪 Running A/B Test: Testing 3 prompt versions...")
print("=" * 60)

sample_texts = [
    "Artificial intelligence is transforming industries worldwide. Machine learning algorithms can now perform tasks that once required human intelligence.",
    "Climate change poses significant environmental challenges. Rising temperatures and extreme weather events impact ecosystems globally.",
    "Blockchain technology enables secure, decentralized transactions. It eliminates the need for intermediaries in many processes.",
]

# Run 30 requests with automatic A/B testing
for i in range(30):
    # Get a prompt version (automatic traffic splitting)
    prompt_version = prompt_manager.get_prompt_for_request(
        "tutorial_summary",
        ab_testing=True
    )
    
    text = random.choice(sample_texts)
    prompt_text = prompt_version.template.format(text=text)
    
    # Simulate LLM call with telemetry
    with tracker.track_request(
        model_name="gpt-3.5-turbo",
        provider="openai",
        prompt_text=prompt_text,
        prompt_id=prompt_version.prompt_id,
        prompt_version=prompt_version.version,
        tags=["tutorial", "ab_test"]
    ) as ctx:
        time.sleep(random.uniform(0.05, 0.15))
        
        # Simulate different quality scores for different versions
        base_quality = {
            1: 0.70,  # Basic version
            2: 0.80,  # Improved version
            3: 0.88   # Expert version
        }.get(prompt_version.version, 0.75)
        
        quality_score = min(0.95, base_quality + random.uniform(-0.05, 0.05))
        
        ctx.set_tokens(
            int(len(prompt_text.split()) * 1.3),
            int(50 * random.uniform(0.8, 1.2))
        )
        ctx.set_response(f"Summary generated by version {prompt_version.version}")
        ctx.set_quality_score(quality_score)
        ctx.set_user_feedback("positive" if quality_score > 0.75 else "negative")
    
    # Update metrics
    prompt_manager.update_prompt_metrics("tutorial_summary", prompt_version.version)
    
    if (i + 1) % 10 == 0:
        print(f"  ✓ Completed {i + 1}/30 A/B test requests")

print("\\n✅ A/B Test completed!")
````

````python
# Analyze A/B test results
print("📊 A/B Test Results:")
print("=" * 60)

versions = prompt_manager.get_prompt_versions("tutorial_summary")
ab_results = []

for version in versions:
    if version.total_calls and version.total_calls > 0:
        ab_results.append({
            'Version': version.version,
            'Name': version.name,
            'Calls': version.total_calls,
            'Avg Latency': f"{version.avg_latency_ms:.0f}ms" if version.avg_latency_ms else "N/A",
            'Avg Cost': f"${version.avg_cost_usd:.6f}" if version.avg_cost_usd else "N/A",
            'Quality Score': f"{version.avg_quality_score:.3f}" if version.avg_quality_score else "N/A",
            'Success Rate': f"{version.success_rate:.1%}" if version.success_rate else "N/A"
        })
        
        print(f"\\nVersion {version.version}: {version.name}")
        print(f"  Group: {version.ab_test_group}")
        print(f"  Calls: {version.total_calls}")
        print(f"  Avg Latency: {version.avg_latency_ms:.0f}ms" if version.avg_latency_ms else "  Avg Latency: N/A")
        print(f"  Quality Score: {version.avg_quality_score:.3f}" if version.avg_quality_score else "  Quality Score: N/A")

# Create comparison DataFrame
df_ab = pd.DataFrame(ab_results)
print("\\n" + df_ab.to_string(index=False))
````

````python
# Visualize A/B test results
if len(ab_results) > 0:
    # Extract numeric quality scores
    quality_data = []
    for v in versions:
        if v.total_calls and v.total_calls > 0 and v.avg_quality_score:
            quality_data.append({
                'Version': f"V{v.version}: {v.name}",
                'Quality Score': v.avg_quality_score,
                'Calls': v.total_calls
            })
    
    if quality_data:
        df_quality = pd.DataFrame(quality_data)
        
        fig = px.bar(
            df_quality,
            x='Version',
            y='Quality Score',
            color='Quality Score',
            title='🎯 A/B Test: Quality Score Comparison',
            text='Quality Score',
            color_continuous_scale='RdYlGn',
            range_color=[0.6, 1.0]
        )
        fig.update_traces(texttemplate='%{text:.3f}', textposition='outside')
        fig.update_layout(showlegend=False, height=400)
        fig.show()
        
        print("\\n💡 Insight: Higher quality scores indicate better prompt performance!")
````

## 🚀 Section 9: Automatic Prompt Optimization

Now comes the magic! The **Automatic Prompt Optimizer** analyzes all collected data and recommends the best-performing prompt variant.

### Optimization Goals:

- **LATENCY**: Minimize response time
- **COST**: Minimize API costs
- **QUALITY**: Maximize response quality
- **BALANCED**: Optimize all metrics together

### How It Works:

1. Collect performance data for all versions
2. Calculate a score based on the optimization goal
3. Recommend the best version
4. Provide detailed metrics for decision-making

````python
# Run automatic optimization
print("🚀 Running Automatic Prompt Optimization...")
print("=" * 60)

result = prompt_optimizer.run_optimization(
    prompt_id="tutorial_summary",
    goal=OptimizationGoal.BALANCED
)

if result["status"] == "success":
    recommendation = result["recommendation"]
    
    print("\\n✅ Optimization Successful!")
    print(f"\\n🎯 Recommended Version: {recommendation['recommended_version']}")
    print(f"   Name: {recommendation['version_name']}")
    print(f"   Optimization Goal: {recommendation['goal']}")
    print(f"   Overall Score: {recommendation['score']:.4f}")
    
    print(f"\\n📊 Performance Metrics:")
    metrics = recommendation['metrics']
    print(f"   Total Calls: {metrics['total_calls']}")
    print(f"   Success Rate: {metrics['success_rate']:.1%}")
    if metrics.get('avg_latency_ms'):
        print(f"   Avg Latency: {metrics['avg_latency_ms']:.0f}ms")
    if metrics.get('avg_cost_usd'):
        print(f"   Avg Cost: ${metrics['avg_cost_usd']:.6f}")
    if metrics.get('avg_quality_score'):
        print(f"   Avg Quality: {metrics['avg_quality_score']:.3f}")
    
    print("\\n💡 Action: Consider making this version the default!")
    
else:
    print(f"⚠️ {result.get('message', 'Optimization could not complete')}")
````

## 💰 Section 10: Cost Analysis and Budget Management

Understanding and managing costs is critical for LLM applications. Let's analyze our spending!

````python
# Cost analysis
print("💰 Cost Analysis Report")
print("=" * 60)

all_models_summary = monitor.get_all_models_summary(time_window_hours=24)

total_cost = 0
total_requests = 0
cost_breakdown = []

for model_stats in all_models_summary:
    if model_stats.get('total_requests', 0) > 0:
        model_cost = model_stats.get('total_cost_usd', 0)
        model_requests = model_stats['total_requests']
        
        total_cost += model_cost
        total_requests += model_requests
        
        cost_breakdown.append({
            'Model': model_stats['model_name'],
            'Requests': model_requests,
            'Total Cost': model_cost,
            'Cost/Request': model_cost / model_requests if model_requests > 0 else 0,
            '% of Budget': 0  # Will calculate after
        })

# Calculate percentages
for item in cost_breakdown:
    item['% of Budget'] = (item['Total Cost'] / total_cost * 100) if total_cost > 0 else 0

print(f"\\n📊 Total Spending: ${total_cost:.6f}")
print(f"📊 Total Requests: {total_requests:,}")
print(f"📊 Average Cost per Request: ${(total_cost / total_requests):.6f}" if total_requests > 0 else "N/A")

print(f"\\n💡 Cost Breakdown by Model:")
for item in cost_breakdown:
    print(f"\\n  {item['Model']}:")
    print(f"    Requests: {item['Requests']:,}")
    print(f"    Total Cost: ${item['Total Cost']:.6f}")
    print(f"    Cost/Request: ${item['Cost/Request']:.6f}")
    print(f"    % of Budget: {item['% of Budget']:.1f}%")
````

````python
# Visualize cost distribution
if cost_breakdown:
    df_cost = pd.DataFrame(cost_breakdown)
    
    fig = px.pie(
        df_cost,
        names='Model',
        values='Total Cost',
        title='💰 Cost Distribution by Model',
        hole=0.4
    )
    fig.update_traces(textposition='inside', textinfo='percent+label')
    fig.show()
    
    print("\\n💡 Insight: Identify which models are consuming your budget!")
````

````python
def complete_workflow_demo():
    """
    Demonstrates the complete end-to-end workflow of the system.
    """
    print("🎯 COMPLETE END-TO-END WORKFLOW DEMO")
    print("=" * 70)
    
    # Step 1: User request comes in
    print("\\n[Step 1] 📥 User Request Received")
    user_input = "Explain the concept of machine learning in simple terms."
    print(f"  Input: '{user_input}'")
    
    # Step 2: Get best prompt version (A/B testing)
    print("\\n[Step 2] 🔄 Selecting Prompt (A/B Testing)")
    prompt_version = prompt_manager.get_prompt_for_request(
        "tutorial_summary",
        ab_testing=True
    )
    print(f"  Selected: Version {prompt_version.version} - {prompt_version.name}")
    print(f"  A/B Group: {prompt_version.ab_test_group}")
    
    # Step 3: Make LLM call with telemetry
    print("\\n[Step 3] 🤖 Making LLM Call (with Telemetry)")
    prompt_text = prompt_version.template.format(text=user_input)
    
    with tracker.track_request(
        model_name="gpt-3.5-turbo",
        provider="openai",
        prompt_text=prompt_text,
        prompt_id=prompt_version.prompt_id,
        prompt_version=prompt_version.version,
        tags=["demo", "end-to-end"]
    ) as ctx:
        start = time.time()
        time.sleep(random.uniform(0.1, 0.2))  # Simulate API call
        elapsed = (time.time() - start) * 1000
        
        response = "Machine learning is a technology that allows computers to learn from data and make decisions without being explicitly programmed for each task."
        
        ctx.set_tokens(
            int(len(prompt_text.split()) * 1.3),
            int(len(response.split()) * 1.3)
        )
        ctx.set_response(response)
        ctx.set_quality_score(0.89)
        ctx.set_user_feedback("positive")
    
    print(f"  ✓ Response received in {elapsed:.0f}ms")
    print(f"  ✓ Telemetry data captured")
    
    # Step 4: Update monitoring metrics
    print("\\n[Step 4] 📊 Updating Monitoring Metrics")
    prompt_manager.update_prompt_metrics("tutorial_summary", prompt_version.version)
    stats = monitor.get_model_stats("gpt-3.5-turbo", time_window_hours=1)
    print(f"  Current Stats:")
    print(f"    - Total Requests: {stats.get('total_requests', 0)}")
    print(f"    - Success Rate: {stats.get('success_rate', 0):.1%}")
    print(f"    - Avg Latency: {stats.get('avg_latency_ms', 0):.0f}ms")
    
    # Step 5: Check alerts
    print("\\n[Step 5] 🚨 Checking Alert Conditions")
    alerts = monitor.get_active_alerts()
    if alerts:
        print(f"  ⚠️ {len(alerts)} active alert(s)")
    else:
        print(f"  ✅ No alerts triggered")
    
    # Step 6: Run optimization check
    print("\\n[Step 6] 🚀 Checking Optimization Opportunities")
    analysis = prompt_optimizer.analyze_prompt_performance("tutorial_summary", 24)
    if analysis.get("recommendation"):
        rec = analysis["recommendation"]
        if rec.get("status") == "recommendation_available":
            print(f"  💡 Recommendation: Use Version {rec['recommended_version']}")
            print(f"     Score: {rec['score']:.4f}")
        else:
            print(f"  ℹ️ {rec.get('message', 'No recommendation yet')}")
    
    # Step 7: Return response to user
    print("\\n[Step 7] 📤 Returning Response to User")
    print(f"  Response: '{response}'")
    
    print("\\n" + "=" * 70)
    print("✅ WORKFLOW COMPLETED SUCCESSFULLY!")
    print("\\n💡 Key Points:")
    print("  • Telemetry was collected automatically")
    print("  • Monitoring provided real-time insights")
    print("  • A/B testing selected the prompt variant")
    print("  • Optimization recommended improvements")
    print("  • All without manual intervention!")

# Run the demo
complete_workflow_demo()
````

## 🎓 Section 14: Key Learnings and Best Practices

### 📚 What We've Learned:

#### 1. **Telemetry Collection**
- ✅ Always track metrics for every LLM call
- ✅ Use context managers for automatic data capture
- ✅ Include metadata and tags for better analysis

#### 2. **Performance Monitoring**
- ✅ Track multiple metrics: latency, cost, quality
- ✅ Use percentiles (P95, P99) for SLA definition
- ✅ Set up alerts for threshold violations

#### 3. **Prompt Engineering**
- ✅ Version control your prompts
- ✅ Test multiple variants simultaneously
- ✅ Let data drive your decisions

#### 4. **Cost Management**
- ✅ Monitor spending in real-time
- ✅ Understand cost per request
- ✅ Optimize for the right model/prompt combination

#### 5. **A/B Testing**
- ✅ Require sufficient sample size (10+ requests minimum)
- ✅ Split traffic evenly or based on weights
- ✅ Use statistical methods for comparison

### 🚀 Production Deployment Tips:

1. **Database**: Use PostgreSQL instead of SQLite for production
2. **Caching**: Cache prompt templates to reduce database queries
3. **Async Processing**: Use async/await for non-blocking operations
4. **Rate Limiting**: Implement rate limiting to control costs
5. **Security**: Encrypt API keys and sensitive data
6. **Monitoring**: Set up alerts for Slack/Email notifications
7. **Backup**: Regular database backups for telemetry data
8. **Retention**: Define data retention policies (e.g., keep 90 days)
